```python
## create a simple txt file

import os
os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_text={
    "../data/text_files/python_intro.txt":"""Python is a high-level, general-purpose programming language celebrated for its simple, English-like syntax that emphasizes human readability. Created by Guido van Rossum and first released in 1991, Python is an open-source language managed by the Python Software Foundation.""",
    "../data/text_files/python_intro2.txt":"""It serves as a versatile tool for building web applications, analyzing complex data, automating repetitive tasks, and powering artificial intelligence systems.Core Technical CharacteristicsInterpreted Nature: Python executes code line-by-line using an interpreter, meaning there is no manual compilation step required before running a program.Dynamically Typed: Variable data types do not need explicit declaration; the system automatically determines them during program runtime.Indentation Rules: Code blocks (like loops or functions) are defined using whitespace indentation instead of traditional curly braces or brackets.Multi-Paradigm Support: Developers can write structured programs using object-oriented, procedural, or functional coding styles.Cross-Platform Portability: A standard script runs natively across Windows, macOS, Linux, and specialized hardware like the Raspberry Pi.Common ApplicationsData Science & AI: Engineers use specialized packages like NumPy and Pandas to perform mathematical operations and analyze large datasets.Machine Learning: Frameworks like TensorFlow and Scikit-learn enable developers to build predictive models and neural networks.Web Development: Backend applications are quickly built using high-speed frameworks such as Django and Flask.Automation & Scripting: System administrators write simple scripts to parse local data files, interact with databases, or scrape online websites."""
}

for filepath,content in sample_text.items():
    with open(filepath, 'w', encoding="utf-8") as f:
        f.write(content)

print("Sample files are created!")
```

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
document = loader.load()
print(document)
```

```python
### Directory Loader
from langchain_community.document_loaders import DirectoryLoader

## load all the text files from the directory
dir_loader=DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt", ## pattern to match
    loader_cls=TextLoader,
    loader_kwargs={'encoding':'utf-8'},
    show_progress=True
)

documents=dir_loader.load()
print(documents)
```

```python
### Directory Loader
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader
from langchain_community.document_loaders import DirectoryLoader

## load all the text files from the directory
dir_loader=DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf", ## pattern to match
    loader_cls=PyMuPDFLoader,
    show_progress=False
)

pdf_documents=dir_loader.load()
pdf_documents
```
